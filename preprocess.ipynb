```python
import io
import pandas as pd
import ipywidgets as widgets
from IPython.display import display

# 1. Load Data


df = pd.read_csv("data.csv")

# 2. Data Inspection
print("--- FIRST 5 ROWS (head) ---")
print(df.head())

print("\n--- DATA FRAME INFO ---")
df.info()

print("\n--- MISSING VALUE COUNT PER COLUMN ---")
print(df.isnull().sum())

# 3. Interactive UI Filters (ipywidgets)
district_options = ["All"] + sorted(df["district"].unique().tolist())
month_options = ["All"] + sorted(df["month"].unique().tolist())
indicator_options = [
    "anc_coverage",
    "institutional_delivery",
    "immunization",
    "high_risk_cases",
]

district_dropdown = widgets.Dropdown(
    options=district_options, value="All", description="District:"
)
month_dropdown = widgets.Dropdown(
    options=month_options, value="All", description="Month:"
)
indicator_dropdown = widgets.Dropdown(
    options=indicator_options,
    value="anc_coverage",
    description="Indicator:",
)

output = widgets.Output()


def filter_data(change):
    output.clear_output()
    filtered_df = df.copy()

    # Filter by District
    if district_dropdown.value != "All":
        filtered_df = filtered_df[
            filtered_df["district"] == district_dropdown.value
        ]

    # Filter by Month
    if month_dropdown.value != "All":
        filtered_df = filtered_df[filtered_df["month"] == month_dropdown.value]

    # Select relevant columns including chosen indicator
    selected_cols = ["month", "district", indicator_dropdown.value]
    result = filtered_df[selected_cols]

    with output:
        display(result)


# Attach observers to dropdown controls
district_dropdown.observe(filter_data, names="value")
month_dropdown.observe(filter_data, names="value")
indicator_dropdown.observe(filter_data, names="value")

# Display UI
print("\n--- INTERACTIVE FILTER UI ---")
display(
    widgets.VBox(
        [
            widgets.HBox(
                [district_dropdown, month_dropdown, indicator_dropdown]
            ),
            output,
        ]
    )
)

# Initial render trigger
filter_data(None)
```
